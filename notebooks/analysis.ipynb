```python
import pandas as pd

df = pd.read_csv("../data/hiit_vs_hypertrophy_gains.csv")

df.head()
```

```python
print("Missing Values")
display(df.isnull().sum())

print("Duplicate rows")
display(df.duplicated().sum())
```

```python
df["Lean_Mass_Change_kg"] = (df["Final_Lean_Mass_kg"] - df["Initial_Lean_Mass_kg"])

display(df["Lean_Mass_Change_kg"])
```

```python
import matplotlib.pyplot as plt

df.boxplot(column="Lean_Mass_Change_kg", by="Group", grid=False)

plt.title("Lean Mass Change by Training Group")
plt.xlabel("Training Group")
plt.ylabel("Lean Mass Change (kg)")
plt.show()
```
